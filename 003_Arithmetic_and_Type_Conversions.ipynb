```python
#Main exercise

meal_price_text = "18.50"
meal_quantity_text = "3"
drink_price_text = "4.50"
drink_quantity_text = "3"

discount_percent = 10
number_of_people = 4

meal_price = float(meal_price_text)
meal_quantity = int(meal_quantity_text)
drink_price = float(drink_price_text)
drink_quantity = int(drink_quantity_text)

meal_total = meal_price * meal_quantity
drink_total = drink_price * drink_quantity

subtotal = meal_total + drink_total
final_total = subtotal - (subtotal * (discount_percent / 100))

cost_per_person = final_total / number_of_people

print("Simple billing calculator")
print("Meal total = ", meal_total)
print("Drink total = ", drink_total)

print("Subtotal = ", subtotal)
print("Final total after discount = ", final_total)
print("Cost per person for split bills = ", cost_per_person)
print("Rounded share:", round(cost_per_person, 2))
```
