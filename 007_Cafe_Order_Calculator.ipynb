1st Project Cafe Order Calculator

```python
#customer_name = ""
customer_name = "  reyhan todo  "
clear_name = customer_name.strip().title()

#product_name = ""
product_name = "  ICED LATTE  "
clear_product = product_name.strip().title()

unit_price = 5
quantity = 3

order_details = (
    clear_name != ""
    and clear_product != ""
    and unit_price > 0
    and 1 <= quantity <= 10
)

is_member = False
has_voucher = True
is_cancelled = False

if is_cancelled:
    print("Order has been cancelled.")
elif not order_details:
    print("Invalid order details.")
    if not (1 <= quantity <= 10):
        print("Quantity must be between 1 and 10.")
else:
    print("Calculating your bills...")
    subtotal = unit_price * quantity
    
    if is_member and subtotal >= 20:
        member_discount = 10
    else:
        member_discount = 0

    discount_amount = subtotal * (member_discount/100)
    final_total = subtotal - discount_amount
    
    if is_member or has_voucher:
        free_drink_msg = "Yes"
    else:
        free_drink_msg = "No"

print("=== CAFE RECEIPT ===")
print(f"Customer : {clear_name}")
print(f"Product : {clear_product}")
print(f"Quantity : {quantity}")
print(f"Unit Price : ${unit_price}")
print(f"Subtotal : {subtotal:.2f}")
print(f"Discount ({member_discount}%) : ${discount_amount:.2f}")
print(f"Total to pay : ${final_total:.2f}")
print(f"Free drink : {free_drink_msg}")
```
