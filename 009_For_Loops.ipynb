```python
products = ["Latte", "Tea", "Croissant"]

for product in products:
    print(f"Available : {product}")

print("Finished displaying the menu.")
```

```python
#Loop with condition
order_totals = [12.0, 25.0, 8.5]

for order_total in order_totals:
    if order_total >= 20:
        print(f"${order_total:.2f}: Large enough for the offer")
    else:
        print(f"${order_total:.2f}: Below the offer minimum")
```

```python
order_totals = [12.0, 25.0, 8.5]

total_sales = 0
number_of_orders = 0

for order_total in order_totals:
    total_sales = total_sales + order_total
    number_of_orders = number_of_orders + 1

print(f"Total sales: ${total_sales:.2f}")
print(f"Number of orders: {number_of_orders}")
```

```python
order_totals = [12.0, 25.0, 8.5, 30.0]
qualifying_orders = []

for order_total in order_totals:
    if order_total >= 20:
        qualifying_orders.append(order_total)

print(qualifying_orders)
```

```python
#Main exercise

order_totals = [19.5, 26.0, 16.5, 42.5, 18.0, 23.4]
qualifying_threshold = 20
qualifying_orders = []
below_treshold = []
sales_target = 150
total_sales = 0
number_of_orders = 0

for order_total in order_totals:
    if order_total >= qualifying_threshold:
        qualifying_orders.append(order_total)
    else:
        below_treshold.append(order_total)
    number_of_orders += 1
    total_sales += order_total

if number_of_orders == 0:
    print("No orders recorded.")
else:
    average_order = total_sales / number_of_orders

    print(f"Number of orders : {number_of_orders}")
    print(f"Total sales : {total_sales:.2f}")
    print(f"Average order : {round(average_order, 2)}")
    print(f"Qualifying orders : {len(qualifying_orders)}")
    print(f"Orders below treshold : {len(below_treshold)}")
    print(f"Qualifying amounts  : {qualifying_orders}")

    if total_sales < sales_target:
        print("Sales target not reached.")
    else:
        print("Sales target reached.")
```

4. First one will summarize each order_total value until loop finished, while second one is for summarize each content inside the list.
