```python
#Main exercise

products = [
    {"name": "  LATTE ", "price": 6.5, "stock": 8},
    {"name": " tea  ", "price": 4.0, "stock": 0},
    {"name": " CROISSANT ", "price": 5.5, "stock": 2},
    {"name": "muffin", "price": 4.5, "stock": 5}
]

low_stock_threshold = 2

out_of_stock = []
low_stock = []
stock_sum = 0
total_value = 0

for product in products:
    cleaned_name = product["name"].strip().title()
    product["name"] = cleaned_name
    stock_value = product["price"] * product["stock"]
    product["stock_value"] = stock_value
    stock_sum += product["stock"]
    total_value += product["stock_value"]
    
    if product["stock"] == 0:
        product["status"] = "Out of stock"
        out_of_stock.append(product["name"])
    elif product["stock"] <= low_stock_threshold:
        product["status"] = "Low Stock"
        low_stock.append(product["name"])
    else:
        product["status"] = "Available"

for product in products:
    print(f"{product["name"]} | Stock : {product["stock"]} | Values : ${product["stock_value"]:.2f} | {product["status"]}")

product_records = len(products)

    
print(f"\nProduct records : {product_records}")
print(f"Total Units : {stock_sum}")
print(f"Total stock value : ${total_value:.2f}")
print(f"Out of stock products: {out_of_stock}")
print(f"Low stock products: {low_stock}")
```
