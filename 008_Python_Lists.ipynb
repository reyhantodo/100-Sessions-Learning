```python
products = ["Latte", "Tea", "Croissant"]
products[1] = "Matcha"

print(products)
```

```python
products = ["Latte", "Tea", "Croissant"]

products.append("Muffin")
print(products)

products.remove("Tea")
print(products)

removed_product = products.pop()
print("Removed:", removed_product)
print("Remaining:", products)
```

```python
products = ["Latte", "Tea", "Croissant"]

print("Tea" in products)          # True
print("Muffin" not in products)   # True
print("latte" in products)        # False
```

```python
requested_product = "  latte "
clean_request = requested_product.strip().title()

if clean_request in products:
    print(f"{clean_request} is on the menu.")
else:
    print(f"{clean_request} is unavailable.")
```

```python
order_totals = [19.5, 26.0, 15.0, 42.5]

number_of_orders = len(order_totals)
total_sales = sum(order_totals)
smallest_order = min(order_totals)
largest_order = max(order_totals)
average_order = total_sales / number_of_orders

print(f"Orders: {number_of_orders}")
print(f"Total sales: ${total_sales:.2f}")
print(f"Smallest order: ${smallest_order:.2f}")
print(f"Largest order: ${largest_order:.2f}")
print(f"Average order: ${average_order:.2f}")
```

```python
#Main exercise

order_totals = [19.5, 26.0, 15.0, 42.5, 18.0]
sales_target = 150

print(order_totals[0], order_totals[4])
print(order_totals[0:3])

order_totals[2] = 16.5
print(order_totals)

order_totals.append(23.4)
print(order_totals)

total_number = len(order_totals)
if len(order_totals) == 0:
    print("No orders recorded.")
else:
    print("\nCalculating Summary")
    total_sales = sum(order_totals)
    min_order = min(order_totals)
    max_order = max(order_totals)
    average_order = total_sales / total_number

    print(f"\nTotal Number: {total_number}")
    print(f"Total sales: ${total_sales:.2f}")
    print(f"Smallest order: ${min_order:.2f}")
    print(f"Largest order: ${max_order:.2f}")
    print(f"Average order: ${average_order:.2f}\n")

    difference_amount = sales_target - total_sales
    print(f"Diffrence amount : ${difference_amount:.2f}")
    if difference_amount > 0:
        print("Sales target not reached.")
    else:
        print("Sales target reached.")
```
