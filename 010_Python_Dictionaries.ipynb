```python
#List

product = ["Latte", 6.5, 12]
```

```python
#Dictionary

product = {
    "name": "Latte",
    "price": 6.5,
    "stock": 12
}

print(product)
print(type(product))
```

```python
print(product["name"])
print(product["price"])
print(product["stock"])
```

```python
quantity = 3
order_total = product["price"] * quantity

print(f"Product: {product['name']}")
print(f"Order total: ${order_total:.2f}")
```

```python
#Update

product["price"] = 7.0
product["stock"] = 10

print(product)
```

```python
product["category"] = "Coffee"
print(product)
```

```python
product["is_available"] = True
print(product)
```

```python
stock_value = product["price"] * product["stock"]
product["stock"] = 5

print(stock_value)
```

```python
#Missing key

print(product["supplier"])
```

```python
if "supplier" in product:
    print(product["supplier"])
else:
    print("Supplier not recorded.")
```

```python
print("name" in product)   # True
print("Latte" in product)  # False
```

```python
supplier = product.get("supplier", "Not recorded")
print(supplier)
```

```python
#Dictionary loop

for key in product:
    print(key)
```

```python
for key, value in product.items():
    print(f"{key}: {value}")
```

```python
#Main exercise

product = {
    "name": "  ICED LATTE  ",
    "price": 6.5,
    "stock": 8
}

requested_quantity = 3

clean_name = product["name"].strip().title()
product["name"] = clean_name

if requested_quantity <= 0:
    print("Quantity must be greater than zero.")
elif requested_quantity > product["stock"]:
    print("Not enough stock.")
else:
    print("Processing the request.")
    
    sales_total = requested_quantity * product["price"]
    product["stock"] = product["stock"] - requested_quantity
    
    print(f"Product : {product["name"]}")
    print(f"Quantity sold : {requested_quantity}")
    print(f"Sale total : ${sales_total:.2f}")
    print(f"Remaining stock : {product["stock"]}")
    
    if product["stock"] <= 2:
        print("Restock needed.")
    else:
        print("Stock level is sufficient.")
```

7. To refresh the memory of program.

```python
#Additional exercise

menu = {
    "Latte": 6.5,
    "Tea": 4.0,
    "Croissant": 5.5
}

for key, value in menu.items():
    print(f"{key}: {value}")

requested_product = "  tea "
clean_product = requested_product.strip().title()

if clean_product in menu:
    selected_price = menu[clean_product]
    print(f"{clean_product} : ${selected_price:.2f}")
else:
    print("Product not on the menu.")
```
