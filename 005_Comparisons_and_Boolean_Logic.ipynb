Comparisons and Boolean Logic

```python
quantity = 3          # Assign a value
print(quantity == 3) # Compare a value
```

```python
order_total = 50
qualifies_for_discount = order_total >= 50

print(qualifies_for_discount)
print(type(qualifies_for_discount))
```

```python
print(10 > 10)
print(10 >= 10)
print(10 != 5)
print("10" == 10)
print("Latte" == "latte")
```

```python
#And condition example

is_member = True
order_total = 35

eligible_for_discount = is_member and (order_total >= 30)

print(eligible_for_discount)

#Or condition example
has_voucher = False

eligible_for_offer = is_member or has_voucher
print(eligible_for_offer)

#Not condition example
is_cancelled = False

print(not is_cancelled)
```

```python
#Boundaries checking example
quantity = 11

quantity_in_range = (quantity >= 1) and (quantity <= 10)
print(quantity_in_range)
```

```python
#Main exercise

customer_name = ""
has_customer_name = customer_name.strip() != ""
print(has_customer_name)

unit_price = 6.5
has_valid_price = unit_price > 0

quantity = 0
has_valid_quantity = 1 <= quantity <= 10

is_valid_order = has_customer_name and has_valid_price and has_valid_quantity

is_member = False
has_voucher = True
is_cancelled = True

subtotal = unit_price * quantity

qualifies_for_discount = is_member and (subtotal >= 20)

qualifies_for_free_drink = is_member or has_voucher
can_process_order = is_valid_order and (not is_cancelled)

print(f"Has customer name : {has_customer_name}")
print(f"Has valid price : {has_valid_price}")
print(f"Has valid quantity : {has_valid_quantity}")
print(f"Is valid order : {is_valid_order}")
print(f"Qualifies for discount : {qualifies_for_discount}")
print(f"Qualifes for free drink : {qualifies_for_free_drink}")
print(f"Can process order : {can_process_order}")
```
