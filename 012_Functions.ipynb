```python
#Main exercise

def clean_product_name(name):
    return name.strip().title()

def calculate_stock_value(price, stock):
    return price * stock

def get_stock_status(stock, threshold = 2):
    if stock == 0:
        return "Out of stock"
    elif stock > 0 and stock <= threshold:
        return "Low stock"
    else:
        return "Available"

print(clean_product_name("  ICED LATTE "))
print(calculate_stock_value(4.0, 0))
print(get_stock_status(3,5))
```

```python
products = [
    {"name": "  LATTE ", "price": 6.5, "stock": 8},
    {"name": " tea  ", "price": 4.0, "stock": 0},
    {"name": " CROISSANT ", "price": 5.5, "stock": 2},
    {"name": "muffin", "price": 4.5, "stock": 5}
]

low_stock_threshold = 2
total_value = 0

for product in products:
    cleaned_name = clean_product_name(product["name"])
    #product["name"] = cleaned_name (no need)
    stock_value = calculate_stock_value(product["price"], product["stock"])
    #product["stock_value"] = stock_value (no need)
    product_status = get_stock_status(product["stock"], low_stock_threshold)
    total_value += stock_value

    print(f"{cleaned_name} | Stock : {product["stock"]} | Values : ${stock_value:.2f} | {product_status}")
    
print(f"\nTotal stock value : ${total_value:.2f}")
```
