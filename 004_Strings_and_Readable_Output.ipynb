```python
#Main exercise

customer_name = "  reyhan todo  "
product_name = "  ICED LATTE "
order_code = "BNE-2048"

unit_price = 6.5
quantity = 3
discount_percent = 20

customer_name_clean = customer_name.title().strip()
product_name_clean = product_name.title().strip()

print(customer_name_clean)
print(product_name_clean)
print(order_code[0:3])
```

```python
subtotal = unit_price * quantity
discount_amount = subtotal * (discount_percent / 100)
final_total = subtotal - discount_amount

print("CAFE RECEIPT")
print(f"Order : {order_code}")
print(f"Location : {order_code[0:3]}")
print(f"Customer : {customer_name_clean}")
print(f"Product : {product_name_clean}")
print(f"Quantity : {quantity}")
print(f"Unit price : ${unit_price:.2f}")
print(f"Subtotal : ${subtotal:.2f}")
print(f"Discount {discount_percent}% : ${discount_amount:.2f}")
print(f"Total to pay : ${final_total:.2f}")
```
